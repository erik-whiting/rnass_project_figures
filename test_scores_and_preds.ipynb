```python
def check_reliability_of_preds(line):
    pred_acc_map = {}
    for m in models:
        pred_key = f"{m}_prediction"
        pred = line[pred_key]
        acc_key = f"{m}_accuracy"
        acc = line[acc_key]

        if pred not in pred_acc_map.keys():
            pred_acc_map[pred] = acc
        else:
            if pred_acc_map[pred] != acc:
                return False
    return True

```

```python
for d in data:
    check = check_reliability_of_preds(d)
    if not check:
        print(f"Problem with {d['datapoint']}")
        break
```
